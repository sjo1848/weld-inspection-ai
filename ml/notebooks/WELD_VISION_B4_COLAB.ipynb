# WELD-VISION-001 — B4.1 validation-only calibration

This notebook prepares the frozen B4 candidate manifest using validation only.
It does **not** evaluate the frozen project test set.


```python
import pathlib
import subprocess

repo = pathlib.Path('/content/weld-inspection-ai')
branch = 'build/mvp-v0.1'

if repo.exists():
    subprocess.run(
        ['git', '-C', str(repo), 'fetch', 'origin', branch],
        check=True,
    )
    subprocess.run(
        ['git', '-C', str(repo), 'switch', branch],
        check=True,
    )
    subprocess.run(
        ['git', '-C', str(repo), 'pull', '--ff-only'],
        check=True,
    )
else:
    subprocess.run(
        [
            'git', 'clone', '--branch', branch, '--single-branch',
            'https://github.com/sjo1848/weld-inspection-ai.git', str(repo),
        ],
        check=True,
    )

subprocess.run(
    ['python3', str(repo / 'ml/scripts/b4_colab_calibrate.py')],
    check=True,
)
```
